```python
# Calcular los conteos y los porcentajes
counts = raw_ranking2023['Tipo Compañia'].value_counts()
percentages = counts / counts.sum() * 100

# Crear el gráfico de barras
ax = counts.plot(
    kind='bar',
    figsize=(10, 6),
    color=colors
)

# Añadir los valores y porcentajes encima de cada barra
for p, percentage in zip(ax.patches, percentages):
    ax.annotate(
        f'{p.get_height()} ({percentage:.1f}%)', 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Frecuencia')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```

```python

# Calcular los porcentajes
percentages = raw_ranking2023['Tipo Compañia'].value_counts(normalize=True) * 100

# Crear el gráfico de barras
ax = percentages.plot(
    kind='bar',
    figsize=(10, 6),
    color=colors_2
)

# Añadir los porcentajes encima de cada barra
for p in ax.patches:
    ax.annotate(
        f'{p.get_height():.1f}%', 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Porcentaje')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```

```python
# Datos de ejemplo
provincias = companias_ecua['Provincia'].value_counts().reset_index()
provincias.columns = ['Provincia', 'Frecuencia']

# Configuración de estilo
sns.set(style="whitegrid")

# Crear el gráfico de barras
plt.figure(figsize=(12, 8))
ax = sns.barplot(x='Provincia', y='Frecuencia', data=provincias, palette="viridis", legend=False)

# Añadir los valores encima de cada barra
for p in ax.patches:
    ax.annotate(
        format(p.get_height(), ',.0f'), 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points',
        fontsize=10,
        color='black'
    )

# Mejorar etiquetas y título
plt.xlabel('Provincia', fontsize=14, labelpad=0)
plt.ylabel('Número de compañias', fontsize=14)
plt.title('Distribución de compañias por Provincia en Ecuador', fontsize=16, fontweight='bold')
plt.xticks(rotation=90, ha='center')


# Mostrar el gráfico
plt.tight_layout()
plt.show()
```
