¿Cuál es la proporción de valores nulos por cada variable?

```python
(
    raw_ranking2022
    .isnull()
    .melt()
    .pipe(
        lambda df: (
            sns.displot(
                data= df,
                y='variable',
                hue='value',
                multiple='fill',
                aspect=2
            )
        )
    )
)
```
