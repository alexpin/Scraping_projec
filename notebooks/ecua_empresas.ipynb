```python
import matplotlib.pyplot as plt

# Crear el gráfico de barras
ax = (
    raw_ranking2022['Tipo Compañia']
    .value_counts()
    .plot(
        kind='bar',
        figsize=(10, 6),
        color='skyblue'
    )
)

# Añadir los valores encima de cada barra
for p in ax.patches:
    ax.annotate(
        str(p.get_height()), 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Frecuencia')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Crear una paleta de colores
colors = sns.color_palette('husl', n_colors=len(raw_ranking2022['Tipo Compañia'].unique()))

# Crear el gráfico de barras
ax = (
    raw_ranking2022['Tipo Compañia']
    .value_counts()
    .plot(
        kind='bar',
        figsize=(10, 6),
        color=colors
    )
)

# Añadir los valores encima de cada barra
for p in ax.patches:
    ax.annotate(
        str(p.get_height()), 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Frecuencia')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Crear una paleta de colores
colors = sns.color_palette('husl', n_colors=len(raw_ranking2022['Tipo Compañia'].unique()))

# Calcular los conteos y los porcentajes
counts = raw_ranking2022['Tipo Compañia'].value_counts()
percentages = counts / counts.sum() * 100

# Crear el gráfico de barras
ax = counts.plot(
    kind='bar',
    figsize=(10, 6),
    color=colors
)

# Añadir los valores y porcentajes encima de cada barra
for p, percentage in zip(ax.patches, percentages):
    ax.annotate(
        f'{p.get_height()} ({percentage:.1f}%)', 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Frecuencia')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Crear una paleta de colores
colors = sns.color_palette('Set3', n_colors=len(raw_ranking2022['Tipo Compañia'].unique()))

# Calcular los porcentajes
percentages = raw_ranking2022['Tipo Compañia'].value_counts(normalize=True) * 100

# Crear el gráfico de barras
ax = percentages.plot(
    kind='bar',
    figsize=(10, 6),
    color=colors
)

# Añadir los porcentajes encima de cada barra
for p in ax.patches:
    ax.annotate(
        f'{p.get_height():.1f}%', 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points'
    )

# Mostrar el gráfico
plt.xlabel('Tipo Compañia')
plt.ylabel('Porcentaje')
plt.title('Distribución de Tipos de Compañia')
plt.show()
```
