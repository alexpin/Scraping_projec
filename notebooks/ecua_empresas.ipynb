```python
import plotly.express as px

# Datos
provincias = raw_ranking2023['Provincia'].value_counts().reset_index()
provincias.columns = ['Provincia', 'Frecuencia']

# Crear el gráfico de barras interactivo con una paleta de colores personalizada
fig = px.bar(provincias, 
             x='Provincia', 
             y='Frecuencia', 
             title='Distribución de compañías por Provincia en Ecuador',
             color='Provincia',  # Añadir color por provincia
             color_discrete_sequence=px.colors.qualitative.Bold)  # Paleta de colores
# Mostrar el gráfico
fig.show()
```

```python
# Datos
provincias = raw_ranking2023['Provincia'].value_counts().reset_index()
provincias.columns = ['Provincia', 'Frecuencia']

# Crear una matriz de datos para el mapa de calor
heatmap_data = provincias.pivot_table(index='Provincia', values='Frecuencia', aggfunc='sum').fillna(0)

# Crear el gráfico de mapa de calor
fig = px.imshow(heatmap_data, 
                labels=dict(x="Provincia", y="Frecuencia", color="Frecuencia"),
                title='Distribución de compañías por Provincia en Ecuador')

# Mostrar el gráfico
fig.show()
```

```python
import plotly.graph_objects as go

# Datos
provincias = raw_ranking2023['Provincia'].value_counts().reset_index()
provincias.columns = ['Provincia', 'Frecuencia']

# Ordenar los datos por frecuencia de menor a mayor
provincias = provincias.sort_values(by='Frecuencia')

# Crear el gráfico de mapa de calor
fig = go.Figure(data=go.Heatmap(
    z=provincias['Frecuencia'],
    x=provincias['Provincia'],
    y=['Frecuencia'] * len(provincias),
    colorscale='Viridis'  # Puedes cambiar la paleta de colores aquí
))

fig.update_layout(
    title='Distribución de compañías por Provincia en Ecuador',
    xaxis_title='Provincia',
    yaxis_title='Frecuencia'
)

# Mostrar el gráfico
fig.show()
```

```python
import plotly.graph_objects as go

# Datos de ejemplo
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

# Crear el gráfico de superficie 3D
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y)])
fig.update_layout(title='Gráfico de Superficie 3D', autosize=False, width=700, height=700)

# Mostrar el gráfico
fig.show()
```
