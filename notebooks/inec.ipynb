```python
value_counts = processed_inec['forma_institucional'].value_counts()

# Crear un DataFrame a partir de los conteos
df_counts = value_counts.reset_index()
df_counts.columns = ['forma_institucional', 'Count']

# Configurar el estilo de seaborn
sns.set(style="whitegrid")

# Crear la gráfica de barras
plt.figure(figsize=(12, 8))
barplot = sns.barplot(x='Count', y='forma_institucional', data=df_counts, palette='viridis')

# Añadir los valores de los conteos en las barras
for index, value in enumerate(df_counts['Count']):
    plt.text(value, index, f'{value:,}', va='center')

# Añadir títulos y etiquetas
plt.title('Distribución de empresas en ecuador segun su forma institucional 2023')
plt.xlabel('Cantidad')
plt.ylabel('Forma Institucional')

# Mostrar la gráfica
plt.show()
```

```python
value_counts = processed_inec['forma_institucional'].value_counts()

# Calcular los porcentajes
percentages = (value_counts / value_counts.sum()) * 100

# Crear un DataFrame a partir de los porcentajes
df_percentages = percentages.reset_index()
df_percentages.columns = ['forma_institucional', 'Percentage']

# Configurar el estilo de seaborn
sns.set(style="whitegrid")

# Crear la gráfica de barras con una paleta de colores
plt.figure(figsize=(12, 8))
barplot = sns.barplot(x='Percentage', y='forma_institucional', data=df_percentages, palette='coolwarm')

# Añadir los valores de los porcentajes en las barras
for index, value in enumerate(df_percentages['Percentage']):
    plt.text(value, index, f'{value:.2f}%', va='center')

# Añadir títulos y etiquetas
plt.title('Distribución de empresas en ecuador segun su forma institucional 2023', fontsize=16, fontweight='bold')
plt.xlabel('Porcentaje', fontsize=14)
plt.ylabel('Forma Institucional', fontsize=14)

# Mostrar la gráfica
plt.show()
```

```python

TotalEmp_csv = r'C:\Users\alexa\OneDrive\Escritorio\Bases_Data\inec\1.1.2.csv'

# Intentar leer el archivo con una codificación específica
try:
    processed_totalEmp = pd.read_csv(TotalEmp_csv, delimiter=';', encoding='utf-8')
except UnicodeDecodeError:
    # Si falla, intentar con otra codificación
    processed_totalEmp = pd.read_csv(TotalEmp_csv, delimiter=';', encoding='latin1')

print(processed_totalEmp)
```

```python
processed_totalEmp.columns = ['año', 'Nro Empresas']

# Configurar el estilo de seaborn
sns.set(style="whitegrid")

# Crear la figura y el gráfico de barras
plt.figure(figsize=(12, 8))
ax = sns.barplot(x='año', y='Nro Empresas', hue='año', data=processed_totalEmp, palette="viridis", legend=False)

# Añadir los valores encima de cada barra
for p in ax.patches:
    ax.annotate(
        format(p.get_height(), ',.0f'), 
        (p.get_x() + p.get_width() / 2., p.get_height()), 
        ha='center', 
        va='center', 
        xytext=(0, 10), 
        textcoords='offset points',
        fontsize=10,
        color='black'
    )

# Mejorar etiquetas y título
plt.xlabel('Año', fontsize=14, labelpad=10)
plt.ylabel('Número de Empresas', fontsize=14)
plt.title('TOTAL EMPRESAS EN ECUADOR, PERÍODO 2012 - 2023', fontsize=16, fontweight='bold')
plt.xticks(rotation=90, ha='center')

# Mostrar el gráfico
plt.tight_layout()
plt.show()
```

```python
# Calcular la línea de tendencia
z = np.polyfit(processed_totalEmp['año'], processed_totalEmp['Nro Empresas'], 1)
p = np.poly1d(z)
plt.plot(processed_totalEmp['año'], p(processed_totalEmp['año']), color='red', linestyle='--', linewidth=2)
```

```python
# Datos de ejemplo
x = processed_totalEmp['año']
y = processed_totalEmp['Nro Empresas']

# Línea Recta
z1 = np.polyfit(x, y, 1)
p1 = np.poly1d(z1)
plt.plot(x, p1(x), linestyle='--', color='red', label='Lineal')

# Polinómica (Grado 2)
z2 = np.polyfit(x, y, 2)
p2 = np.poly1d(z2)
plt.plot(x, p2(x), linestyle='--', color='blue', label='Polinómica (Grado 2)')

# Exponencial
z3 = np.polyfit(x, np.log(y), 1)
p3 = np.poly1d(z3)
plt.plot(x, np.exp(p3(x)), linestyle='--', color='green', label='Exponencial')

# Logarítmica
z4 = np.polyfit(np.log(x), y, 1)
p4 = np.poly1d(z4)
plt.plot(x, p4(np.log(x)), linestyle='--', color='purple', label='Logarítmica')

# Media Móvil
processed_totalEmp['Media Móvil'] = processed_totalEmp['Nro Empresas'].rolling(window=3).mean()
plt.plot(processed_totalEmp['año'], processed_totalEmp['Media Móvil'], linestyle='--', color='orange', label='Media Móvil')

# Añadir títulos y etiquetas
plt.title('Líneas de Tendencia')
plt.xlabel('Año')
plt.ylabel('Número de Empresas')
plt.legend()

# Mostrar la gráfica
plt.show()
```

```python
processed_totalEmp.columns = ['año', 'Nro Empresas']

# Calcular los porcentajes de cambio año a año
processed_totalEmp['Porcentaje Cambio'] = processed_totalEmp['Nro Empresas'].pct_change() * 100

# Configurar el estilo de seaborn
sns.set(style="whitegrid")

# Crear la gráfica de líneas
plt.figure(figsize=(12, 8))
lineplot = sns.lineplot(x='año', y='Nro Empresas', data=processed_totalEmp, marker='o', color='b')

# Añadir los valores de los porcentajes de cambio en las líneas
for i in range(1, len(processed_totalEmp)):
    plt.text(processed_totalEmp['año'][i], processed_totalEmp['Nro Empresas'][i], 
             f'{processed_totalEmp["Porcentaje Cambio"][i]:.2f}%', 
             horizontalalignment='left', size='medium', color='black', weight='semibold')

# Añadir títulos y etiquetas
plt.title('Evolución porcentual del total de Empresas', fontsize=16, fontweight='bold')
plt.xlabel('Año', fontsize=16)
plt.ylabel('Total de Empresas', fontsize=16)

# Mostrar la gráfica
plt.show()
```

```python
processed_totalEmp.columns = ['año', 'Nro Empresas']

# Calcular la media móvil
processed_totalEmp['Media Móvil'] = processed_totalEmp['Nro Empresas'].rolling(window=2).mean()

# Configurar el estilo de seaborn
sns.set(style="whitegrid")

# Crear la figura y graficar la media móvil
plt.figure(figsize=(12, 8))
plt.plot(processed_totalEmp['año'], processed_totalEmp['Media Móvil'], linestyle='--', color='orange', label='Media Móvil')

# Etiquetas y título
plt.xlabel('Año', fontsize=14, labelpad=10)
plt.ylabel('Número de Empresas', fontsize=14)
plt.title('Media Móvil del total de Empresas 2012 - 2023', fontsize=16, fontweight='bold')
plt.xticks(rotation=90, ha='center')
plt.legend()

# Mostrar el gráfico
plt.tight_layout()
plt.show()
```
