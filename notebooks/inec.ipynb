Cargar dataframe

```python
# Ajustar la configuración para mostrar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
empresas_csv = r'C:\Users\alexa\OneDrive\Escritorio\Bases_Data\inec\EMPRESAS2023_periodo_2023.csv'
raw_inec = pd.read_csv(empresas_csv, delimiter=';')
```

```python
# Identificar valores únicos en la columna
print(raw_inec['ventas_nacionales'].unique())
print(raw_inec['ventas_totales'].unique())
print(raw_inec['propietarios_sexo'].unique())
```

```python
# Lista de columnas a convertir
columns_to_convert = ['ventas_totales', 'ventas_nacionales', 'exportaciones_netas',
                      'plazas_equiv', 'plazas_equiv_hombres', 'plazas_equiv_mujeres',
                      'plazas', 'plazas_hombres', 'plazas_mujeres', 'plazas_ult']

# Convertir cada columna a numérico y luego a Int64
for column in columns_to_convert:
    raw_inec[column] = pd.to_numeric(raw_inec[column], errors='coerce')
    raw_inec[column] = raw_inec[column].astype('Int64')

# Verificar la conversión
print(raw_inec.dtypes)
```
